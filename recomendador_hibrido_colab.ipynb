# Recomendador Híbrido (Content-Based + Colaborativo)

Proyecto de sistema de recomendación listo para **Google Colab**.

**Pregunta:** ¿cómo combinar content-based y filtrado colaborativo?

**Método:** Promedio ponderado de scores

**Stack:** pandas, scikit-learn

Datos de ejemplo.

## 0. Setup

```python
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
print("OK")
```

## 1. Datos de ejemplo

```python
items = pd.DataFrame({
    "id": [1, 2, 3, 4, 5],
    "titulo": ["Item A", "Item B", "Item C", "Item D", "Item E"],
    "descripcion": ["accion aventura", "comedia romance", "accion drama", "romance drama", "aventura comedia"]
})

# Ratings simulados (colaborativo)
ratings = pd.DataFrame({
    "user": [1, 1, 1, 2, 2],
    "item": [1, 2, 3, 1, 4],
    "rating": [5, 3, 4, 4, 5]
})
print(items)
print(ratings)
```

## 2. Content-Based (TF-IDF)

```python
tfidf = TfidfVectorizer()
tfidf_matrix = tfidf.fit_transform(items["descripcion"])
content_sim = cosine_similarity(tfidf_matrix)
print("Content similarity listo")
```

## 3. Colaborativo (promedio simple de ratings)

```python
item_avg = ratings.groupby("item")["rating"].mean().to_dict()
print(item_avg)
```

## 4. Score híbrido

```python
def score_hibrido(item_id, content_weight=0.5):
    content_score = content_sim[item_id-1].mean()
    collab_score = item_avg.get(item_id, 3) / 5.0  # normalizar
    return content_weight * content_score + (1 - content_weight) * collab_score

scores = {i: score_hibrido(i) for i in items["id"]}
print(sorted(scores.items(), key=lambda x: x[1], reverse=True))
```

## 5. Conclusiones

- El enfoque híbrido combina fortalezas de ambos métodos.
- Content-based ayuda con cold-start, colaborativo con preferencias reales.
- Se puede mejorar con modelos más sofisticados (SVD + embeddings).
